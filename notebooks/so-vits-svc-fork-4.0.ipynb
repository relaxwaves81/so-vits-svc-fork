```python
#@title audio augmentation for single audio file if needed
import librosa
import numpy as np
import soundfile as sf
from pydub import AudioSegment
from io import BytesIO
import pyloudnorm as pyln  # 用于音频标准化
import scipy.signal as signal

# 读取 MP3 文件并转换为 WAV
def load_audio_mp3(file_path, sr=44100):
    # 使用pydub加载MP3并转换为WAV
    audio = AudioSegment.from_mp3(file_path)
    audio = audio.set_channels(1)  # 转为单声道
    audio = audio.set_frame_rate(sr)  # 设置采样率
    # 转换为 NumPy 数组并归一化
    samples = np.array(audio.get_array_of_samples(), dtype=np.float32)
    samples = samples / np.iinfo(np.int16).max  # 归一化到 [-1, 1] 范围
    return samples, sr

# 保存音频文件为 MP3 格式
def save_audio_mp3(file_path, y, sr=44100):
    # 确保数据在 [-1, 1] 范围内
    y = np.clip(y, -1.0, 1.0)
    # 转换回 int16 格式
    y_int16 = (y * np.iinfo(np.int16).max).astype(np.int16)
    # 创建 AudioSegment 对象
    audio = AudioSegment(
        y_int16.tobytes(),
        frame_rate=sr,
        sample_width=2,  # 16-bit audio
        channels=1
    )
    # 导出为 MP3
    audio.export(file_path, format="mp3")

# 音调变换：调整音高（更细致的变化）
def pitch_shift(y, sr, n_steps):
    # 添加淡入淡出以避免爆音
    fade_length = int(sr * 0.01)  # 10ms淡变
    y = np.concatenate([np.linspace(0, 1, fade_length) * y[:fade_length],
                       y[fade_length:-fade_length],
                       np.linspace(1, 0, fade_length) * y[-fade_length:]])
    return librosa.effects.pitch_shift(y=y, sr=sr, n_steps=n_steps)

# 变速：调整语速（更小的变化幅度）
def time_stretch(y, rate):
    return librosa.effects.time_stretch(y=y, rate=rate)

# 频谱增强：添加轻微的频谱扭曲
def spec_augment(y, sr):
    # 计算梅尔频谱
    mel_spec = librosa.feature.melspectrogram(y=y, sr=sr)
    # 添加轻微的频谱扭曲
    freq_mask = np.random.uniform(0.8, 1.2, mel_spec.shape[0])
    mel_spec = mel_spec * freq_mask[:, np.newaxis]
    # 转回音频
    return librosa.feature.inverse.mel_to_audio(mel_spec, sr=sr)

# 声码器重建
def vocoder_rebuild(y, sr):
    # 使用 Griffin-Lim 算法重建
    S = np.abs(librosa.stft(y))
    y_rebuilt = librosa.griffin_lim(S)
    return y_rebuilt

# 添加房间混响效果
def add_reverb(y, sr):
    reverb_length = int(sr * 0.1)  # 100ms混响
    impulse_response = np.exp(-np.linspace(0, 5, reverb_length))
    return signal.convolve(y, impulse_response, mode='same')

# 标准化音量
def normalize_audio(y):
    meter = pyln.Meter(44100)  # 创建响度计量器
    loudness = meter.integrated_loudness(y)
    target_loudness = -23.0  # LUFS标准值
    return pyln.normalize.loudness(y, loudness, target_loudness)

# 增强音频并保存
def augment_audio(file_path, output_dir):
    print(f"Loading audio file: {file_path}")
    y, sr = load_audio_mp3(file_path, sr=44100)

    # 先保存原始音频的副本
    for i in range(3):  # 保存3份原始音频副本
        output_path = f"{output_dir}/original_copy_{i}.mp3"
        save_audio_mp3(output_path, y, sr)
        print(f"Saved original copy: {output_path}")

    # 标准化原始音频
    y = normalize_audio(y)

    augmentations = {
        # 音调变化（更细致的变化）
        'pitch_up_1': lambda: pitch_shift(y, sr, 1),
        #'pitch_up_2': lambda: pitch_shift(y, sr, 2),
        'pitch_down_1': lambda: pitch_shift(y, sr, -1),
        #'pitch_down_2': lambda: pitch_shift(y, sr, -2),

        # 语速变化（更小的变化幅度）
        'speed_0.95': lambda: time_stretch(y, 0.95),
        'speed_1.05': lambda: time_stretch(y, 1.05),
        #'speed_0.90': lambda: time_stretch(y, 0.90),
        #'speed_1.10': lambda: time_stretch(y, 1.10),

        # 声码器重建
        #'vocoder_rebuild': lambda: vocoder_rebuild(y, sr),

        # 频谱增强
        'spec_augment_1': lambda: spec_augment(y, sr),
        #'spec_augment_2': lambda: spec_augment(y, sr),  # 随机性会产生不同结果

        # 混响效果
        #'reverb': lambda: add_reverb(y, sr),

        # 组合增强
        'combo_1': lambda: pitch_shift(time_stretch(y, 1.05), sr, 1)
        #'combo_2': lambda: pitch_shift(time_stretch(y, 0.95), sr, -1)
    }

    # 质量控制参数
    min_duration = len(y) * 0.9  # 最小持续时间
    max_duration = len(y) * 1.1  # 最大持续时间

    for name, aug_func in augmentations.items():
        try:
            print(f"Processing {name} augmentation...")
            augmented = aug_func()

            # 质量控制检查
            if len(augmented) < min_duration or len(augmented) > max_duration:
                print(f"Warning: {name} duration out of acceptable range")
                continue

            if np.max(np.abs(augmented)) > 1.0:
                print(f"Warning: {name} contains clipping")
                augmented = np.clip(augmented, -1.0, 1.0)

            # 标准化增强后的音频
            augmented = normalize_audio(augmented)

            output_path = f"{output_dir}/{name}.mp3"
            save_audio_mp3(output_path, augmented, sr)
            print(f"Saved {output_path}")

        except Exception as e:
            print(f"Error processing {name}: {str(e)}")

# 示例用法
if __name__ == "__main__":
    DATASET_NAME = "luning10s"  #@param {type: "string"}
    ORIGINAL_MP3_FILENAME = "original.mp3"  #@param {type: "string"}
    database_folder = f"/content/drive/MyDrive/so-vits-svc-fork/dataset/{DATASET_NAME}/"
    input_file = database_folder + ORIGINAL_MP3_FILENAME
    output_directory = database_folder
    augment_audio(input_file, output_directory)
```
